```r

1 - pchisq(5.51, df = 2) #função de distribuição acumulada da qui-quadrado
#pchisq(q=vetor de quantis, df = graus de liberdade)

x = rnorm(1000) #gera números aleatórios da distribuição normal,neste caso
#neste caso são 1000 valores e uma normal padrão
hist(x, freq = FALSE) #histograma, freq = FALSE: área total sob o gráfico = 1
curve( dnorm(x), add = TRUE) #dnorm = densidade da curva normal(0,1),
#add=TRUE: adiciona a curva ao gráfico anterior

#Metodo da tranformação inversa

U = runif(1000) #gera um vetor de 1000 valores aleatórios de uma uniforme(0,1)
x = qnorm(U) #aplica a função quantil inversa da distribuição normal
#a cada probabilidade em U. Isso mapeia as probabilidades uniformes para
#valores na escala da curva normal (média 0, desvio padrão 1).
hist(x, freq = FALSE) #histograma, freq = FALSE: área total sob o gráfico = 1
curve( dnorm(x), add = TRUE) #dnorm = densidade da curva normal(0,1),
#add=TRUE: adiciona a curva ao gráfico anterior


# Metodo da tranformação inversa

U = runif(1000) #gera um vetor de 1000 valores aleatórios de uma uniforme(0,1)
x = qexp(U, rate = 1/2) #aplica a função quantil inversa de uma
#exponencial(lambda=0,5) a cada probabilidade em U. Isso mapeia as probabilidades
#uniformes para valores na escala da exponencial(média 2).
hist(x, freq = FALSE) #histograma, freq = FALSE: escala o eixo Y para representar
#densidade de probabilidade em vez de contagem bruta
curve( dexp(x, rate = 1/2), add = TRUE) #dnorm = densidade de exp(lambda=0,5),
#add=TRUE: adiciona a curva ao gráfico anterior


# Gerando da gamma(3,2)
n = 1000
u = runif(n) #gera um vetor de n valores aleatórios de uma uniforme(0,1)
x = rgamma(n, shape = 3, rate = 2) #gera uma gama aleatória com forma 3 e taxa 2
#a média teórica é forma/taxa = 3/2 = 1,5
hist(x, freq = FALSE) #plota o histograma da densidade, com área total = 1
curve(dgamma(x, shape = 3, rate = 2), add = TRUE) #dgamma = densidade da gama
#de parâmetros (forma = 3, taxa = 2)
y = qgamma(u, shape = 3, rate = 2) #u=runif(0,1); qgama = quantil inversa
# Ela pega cada probabilidade em u e calcula o valor correspondente
#na escala da distribuição Gama(3,2);
#o vetor y tem 1000 números aleatórios que seguem Gama(3,2)
hist(y, freq = FALSE) #ao invés de mostrar a contagem absoluta, o freq = FALSE
#mostra a densidade de probabilidade, onde a área total é igual a 1
curve(dgamma(x, shape = 3, rate = 2), add = TRUE) #densidade da gamma
#com parâmetros(forma=3, taxa=2), add = TRUE adiciona a curva no gráfico anterior


#Usando soma de 3 exponenciais(2) iids
n = 1000
y = rep (0 , n ) #repete o valor 0 n vezes, neste caso, 1000
for ( i in 1:n ) {
  u = runif (3 , 0 , 1) #gera 3 numeros aleatorios da uniforme(0,1)
  y [ i ] = sum ( -1 / 2 * log ( u ) ) #transforma os valores uniforme para uma
#exp(1)=Gama(1,1). Multiplicado por 1/2, transforma em Gama(1,2). Somando as 3
#Gama dá uma Gama(3,2), que é igual a uma Qui-Quadrado com 6 gl. Depois armazena
#no vetor y[i].
}
hist(y, freq = FALSE) #plota histograma, com área total da densidade de prob = 1
curve(dgamma(x, shape = 3, rate = 2), add = TRUE) #adiciona a curva de densidade
#de probabilidade Gama(3,2) ao gráfico anterior.


#Mistura de normais
n = 500 #gera 500 observações
mu = 10 #média do primeiro grupo
eta = 0.2 #probabilidade de uma observação pertencer ao segundo grupo, ié, 20%
a = 5 #deslocamento da média do segundo grupo
#gerei duas Normais, uma de média 10 e variância 1, outra de média 5 e variância 1
x = y = rep (0 , n ) #Aqui ele cria dois vetores com 500 zeros:
#y → vai indicar qual grupo cada observação pertence.
#x → vai guardar os valores observados.
for ( i in 1: n ) { #Vai repetir o processo 500 vezes, uma para cada observação.
  y [ i ] = rbinom (1 , 1 , eta ) #rbinom(1, 1, 0.2) sorteia um valor de uma Bernoulli:
#0 com probabilidade 0.8; 1 com probabilidade 0.2
#Então, aproximadamente: 80% das vezes → y[i] = 0; 20% das vezes → y[i] = 1
  x [ i ] = rnorm (1 , mean = mu #agora ele gera uma normal dependendo de y[i]
                   + a * y [ i ] , sd = 1) #se y[i]=0, então mu + a*y[i] =
#= 10 + 5*(0) = 10, logo, X[i] é uma N(10,1). Mas se y[i] = 1, X[i] é uma N(15,1)
}
hist(x, freq = FALSE, xlim= c(8,17), ylim= c(0,0.40)) #freq = FALSE faz representar a
#densidade e não a contagem. Isso mostra o histograma de 500 valores gerados.
curve( (1-eta) * dnorm(x, mean = mu, sd = 1) + eta * dnorm(x, mean = mu + a, sd = 1),
       add = TRUE)
#Essa é a densidade teórica da mistura: f(x)=(1-\eta)\phi(x;10,1)+\eta\phi(x;15,1)
#Substituindo os valores: f(x)=0.8ϕ(x;10,1)+0.2ϕ(x;15,1)
#Ou seja: 80% da densidade vem da normal com média 10 + 20% vem da normal com média 15.

# Gerando da t-Student(v)
n = 500 #gera 500 observações pra x
#Ao invés de pegar 500 valores já da t-student, o código constrói esses valores através
#de 2 etapas aleatórias.
nu = 2 #O nu representa o v, que são os graus de liberdade,ié, uma t com apenas 2 graus
#de liberdade, que tem caudas bem pesadas para sua densidade.
sig2 = x = 0 #Inicialmente temos x=0 e sigma²=0, mas isso é só uma inicialização,
#pois quando o R começar, ele preencherá as posições dos dois vetores de tamanho 500.
for ( i in 1: n ) { #para i entre 1 e 500...
  sig2 [ i ] = 1 / rgamma (1 , nu / 2 , nu/ 2) #aqui temos 1/valor aleatório de uma
#Gama(1,1,1) inicialmente. Gama(nº de valores, shape, rate).
#Temos, em resumo, sigma² = 1/Gama[i], com Gama[i] ~ Gama(v/2,v/2)
#No início de cada código sorteia-se um valor aleatório para sigma²
  x [ i ] = rnorm (1 ,0 , sd = sqrt (sig2 [ i ]) ) #Aqui ele gera uma observação de uma
#Normal. Normal(nº de observações, média, variância), mas o rnorm pede desvio-padrão
#e não variância, por isso a troca da variável.
} #Em resumo: GAMA -> INVERTE -> SIGMA² -> NORMAL -> X
hist(x, freq = FALSE) #Cria o histograma e faz o Y representar densidade
curve(dt(x, df = nu), add = TRUE ) #Adiciona a curva de densidade da t no histograma.
curve(dnorm(x), add = TRUE, col = 2 ) #Adiciona a curva de densidade da Normal junto.
#Caudas da t são mais pesadas que as caudas da Normal, mas o pico da Normal é maior.
#De fato, relacionando as duas: Pico maior = caudas leves. Pico menor = caudas pesadas.

# Rejeicao

# Ex1 Amostrar de N(0,1) usando unif(-10, 10) como envelope
p = function(theta){
  return( dnorm(theta) ) #Definiu uma função de theta e p como densidade da Normal
}
q = function(theta){
  return( 1/20* I(-10 < theta)*I(theta<10) )
} #Quando theta está entre -10 e 10, neste caso, da Unif(-10,10), retorna 0,05
#Quando não está, o valor numérico é 0(FALSE) e não é levado em conta na densidade
theta = rep(0, 500) #Guarda os valores criando um vetor de 500 parâmetros
i = 1
n_aceita = 0 #Não aceitamos nenhum ainda, está definido como 0 o contador.
while (n_aceita < 500){ #Escolhe aleatoriamente até ter 500 valores aceitos
  theta_s = runif(1, -10, 10) #Gera de 1 em 1 valores da unif(-10,10)
  A = 20/sqrt(2*pi) #Queremos achar o valor da constante A t.q.
  #p(theta) <= A*q(theta)
  R = p(theta_s)/(A * q(theta_s)) #Isso define R = exp{-theta²/2}
  #É interessante, pois quanto mais próximo de 0, maior a probabilidade do valor ser
  #aceito, sendo semelhante à distribuição Normal Padrão.
  moeda = runif(1) #Sorteia um valor aleatório entre 0 e 1
  if( moeda < R ){
    n_aceita = n_aceita + 1
    theta[n_aceita] = theta_s
  }# Se o valor for menor que R, é aceito e atualiza-se o número e valor do candidato
}

hist(theta, freq = FALSE) #Faz o histograma dos vetores que guardamos, sendo aceitos.
curve(dnorm(x), add = TRUE) #Comparamos com a Normal
```
