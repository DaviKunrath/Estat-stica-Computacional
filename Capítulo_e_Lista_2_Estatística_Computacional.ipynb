# Slice Sampler

```r
# Gerando da N(0,1) truncada em (0, 1)
p = function(x){
  dnorm(x) / ( pnorm(1) - pnorm(0) )
}

calculate_cx = function(y){
  sqrt( - 2*log( sqrt(2*pi)*(pnorm(1) - pnorm(0)) * y ) )
}

# inicialize u
u = runif(1, 0, p(0))

n_iter = 1000
chain_x = rep(0, n_iter)
chain_u = rep(0, n_iter)

for (iter in 1:n_iter){
  # amostre x | u
  cx = calculate_cx(y = u)
  limite_direita = min( c(cx, 1) )
  x = runif(1, 0, limite_direita)

  # amostre u|x
  u = runif(1, 0, p(x) )

  # armazene nas cadeias os valores amostrados na itercao i
  chain_x[iter] = x
  chain_u[iter] = u

  print(iter)
}

plot(x = chain_x, y = chain_u, xlim = c(-1, 2), col = "gray" )
curve(p(x), add = TRUE, xlim = c(0,1), col = "blue", lwd = 2 )

hist(chain_x, freq = FALSE)
curve(p(x), add = TRUE, xlim = c(0,1), col = "blue", lwd = 2 )

plot(chain_x, type = "l")
plot(chain_u, type = "l")
acf(chain_x)
acf(chain_u)

plot(x = chain_x[1:100], y = chain_u[1:100], xlim = c(-1, 2), col = "gray", type = "l" )
```
